# Day 3: Stock Market Dashboard

Visualization-only notebook. First run [Day 3 Stock Trends](../../daily_projects/day_03_stock_trends.ipynb) to generate `daily_projects/data/day_03_stock.csv`.

Charts: price and moving averages, return distribution, and annual returns.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

root = Path.cwd()
if (root / "daily_projects").is_dir():
    root = root / "daily_projects"
elif (root.parent / "daily_projects").is_dir():
    root = root.parent / "daily_projects"
elif (root.parent.parent / "daily_projects").is_dir():
    root = root.parent.parent / "daily_projects"
stock = pd.read_csv(root / "data" / "day_03_stock.csv", parse_dates=["date"]).set_index("date")
annual_returns = stock.close.resample("YE").last().pct_change().dropna() * 100
colors = ["#126e82", "#e07a5f", "#3d9970", "#596b9a"]
print(f"Loaded {len(stock):,} daily observations | Latest close: ${stock.close.iloc[-1]:.2f}")
fig, axes = plt.subplots(2, 2, figsize=(13, 9), constrained_layout=True)
fig.suptitle("Day 3 | Market Trend Monitor", fontsize=16, fontweight="bold")
axes = axes.ravel()
axes[0].plot(stock.index, stock.close, color=colors[0], label="Close", linewidth=1.5)
axes[0].plot(stock.index, stock.ma_20, color=colors[1], label="20-day MA")
axes[0].plot(stock.index, stock.ma_50, color=colors[2], label="50-day MA")
axes[0].set(title="Price and moving averages", ylabel="Price ($)")
axes[0].legend(frameon=False)
axes[1].hist(stock.daily_return * 100, bins=35, color=colors[3], edgecolor="white")
axes[1].set(title="Daily return distribution", xlabel="Return (%)", ylabel="Trading days")
axes[2].bar(annual_returns.index.year.astype(str), annual_returns.values, color=colors[2])
axes[2].axhline(0, color="#253247", linewidth=0.8)
axes[2].set(title="Calendar-year return", ylabel="Return (%)")
axes[3].axis("off")
for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)
plt.show()
```
