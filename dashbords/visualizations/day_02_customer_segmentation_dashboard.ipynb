# Day 2: Customer Segmentation Dashboard

Visualization-only notebook. First run [Day 2 Customer Segmentation](../../daily_projects/day_02_customer_segmentation.ipynb) to generate `daily_projects/data/day_02_customers.csv`.

Charts: customer clusters, segment distribution, and average segment spend.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

root = Path.cwd()
if (root / "daily_projects").is_dir():
    root = root / "daily_projects"
elif (root.parent / "daily_projects").is_dir():
    root = root.parent / "daily_projects"
elif (root.parent.parent / "daily_projects").is_dir():
    root = root.parent.parent / "daily_projects"
customers = pd.read_csv(root / "data" / "day_02_customers.csv")
profile = customers.groupby("segment").agg(customers=("spend", "size"), average_spend=("spend", "mean"), average_orders=("orders", "mean"))
colors = ["#126e82", "#e07a5f", "#3d9970", "#e3a72f", "#596b9a"]
print(f"Customers: {len(customers):,} | Segments: {customers.segment.nunique()}")
fig, axes = plt.subplots(2, 2, figsize=(13, 9), constrained_layout=True)
fig.suptitle("Day 2 | Customer Segmentation", fontsize=16, fontweight="bold")
axes = axes.ravel()
for segment, group in customers.groupby("segment"):
    axes[0].scatter(group.orders, group.spend, s=24, alpha=0.65, color=colors[int(segment)], label=f"Segment {segment}")
axes[0].set(title="Spend vs. order frequency", xlabel="Orders", ylabel="Total spend ($)")
axes[0].legend(frameon=False)
axes[1].pie(profile.customers, labels=[f"Segment {i}" for i in profile.index], autopct="%1.0f%%", colors=[colors[int(i)] for i in profile.index], startangle=90)
axes[1].set(title="Customer distribution")
axes[2].bar([f"Segment {i}" for i in profile.index], profile.average_spend, color=[colors[int(i)] for i in profile.index])
axes[2].set(title="Average spend by segment", ylabel="Spend ($)")
axes[3].axis("off")
for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)
plt.show()
```
