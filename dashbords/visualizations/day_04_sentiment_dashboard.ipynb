# Day 4: Sentiment Dashboard

Visualization-only notebook. First run [Day 4 Sentiment Analysis](../../daily_projects/day_04_sentiment_analysis.ipynb) to generate `daily_projects/data/day_04_posts.csv`.

Charts: sentiment mix, frequent terms, and engagement by sentiment. Posts are synthetic; sentiment uses a small lexicon.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

root = Path.cwd()
if (root / "daily_projects").is_dir():
    root = root / "daily_projects"
elif (root.parent / "daily_projects").is_dir():
    root = root.parent / "daily_projects"
elif (root.parent.parent / "daily_projects").is_dir():
    root = root.parent.parent / "daily_projects"
posts = pd.read_csv(root / "data" / "day_04_posts.csv")
counts = posts.sentiment.value_counts().reindex(["Positive", "Neutral", "Negative"])
terms = posts.text.str.lower().str.replace(r"[^a-z ]", "", regex=True).str.split().explode()
stopwords = {"the", "and", "is", "to", "it", "was", "i", "this", "for", "a", "of", "in", "my", "am"}
term_counts = terms[~terms.isin(stopwords)].value_counts().head(12).sort_values()
engagement = posts.groupby("sentiment").engagement.mean().reindex(["Positive", "Neutral", "Negative"])
colors = ["#126e82", "#e07a5f", "#3d9970", "#e3a72f"]
print(f"Loaded {len(posts):,} synthetic posts")
fig, axes = plt.subplots(2, 2, figsize=(13, 9), constrained_layout=True)
fig.suptitle("Day 4 | Social Listening Snapshot", fontsize=16, fontweight="bold")
axes = axes.ravel()
axes[0].bar(counts.index, counts.values, color=["#3d9970", colors[3], colors[1]])
axes[0].set(title="Lexicon sentiment mix", ylabel="Posts")
axes[1].barh(term_counts.index, term_counts.values, color=colors[0])
axes[1].set(title="Frequent terms", xlabel="Mentions")
axes[2].bar(engagement.index, engagement.values, color=["#3d9970", colors[3], colors[1]])
axes[2].set(title="Average engagement by sentiment", ylabel="Engagement")
axes[3].axis("off")
for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)
plt.show()
```
