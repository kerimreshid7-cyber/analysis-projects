# Day 1: Sales Dashboard

Visualization-only notebook. First run [Day 1 Sales Analysis](../../daily_projects/day_01_sales_analysis.ipynb) to generate `daily_projects/data/day_01_sales.csv`.

Charts: monthly revenue, category revenue, and regional revenue.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

root = Path.cwd()
if (root / "daily_projects").is_dir():
    root = root / "daily_projects"
elif (root.parent / "daily_projects").is_dir():
    root = root.parent / "daily_projects"
elif (root.parent.parent / "daily_projects").is_dir():
    root = root.parent.parent / "daily_projects"
sales = pd.read_csv(root / "data" / "day_01_sales.csv", parse_dates=["date", "month"])
monthly = sales.groupby("month").revenue.sum()
category = sales.groupby("category").revenue.sum().sort_values()
region = sales.groupby("region").revenue.sum().sort_values()
colors = ["#126e82", "#e07a5f", "#3d9970", "#e3a72f"]
print(f"Orders: {len(sales):,} | Revenue: ${sales.revenue.sum():,.0f} | Average order: ${sales.revenue.mean():,.2f}")
fig, axes = plt.subplots(2, 2, figsize=(13, 8), constrained_layout=True)
fig.suptitle("Day 1 | Sales Performance", fontsize=16, fontweight="bold")
axes = axes.ravel()
axes[0].plot(monthly.index, monthly.values, color=colors[0], linewidth=2.4, marker="o", markersize=3)
axes[0].set(title="Monthly revenue", ylabel="Revenue ($)")
axes[0].tick_params(axis="x", rotation=35)
axes[1].bar(category.index, category.values, color=colors)
axes[1].set(title="Revenue by category", ylabel="Revenue ($)")
axes[1].tick_params(axis="x", rotation=20)
axes[2].barh(region.index, region.values, color=colors[2])
axes[2].set(title="Revenue by region", xlabel="Revenue ($)")
axes[3].axis("off")
for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)
plt.show()
```
